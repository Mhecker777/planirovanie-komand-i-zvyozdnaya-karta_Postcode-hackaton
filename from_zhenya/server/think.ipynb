```python
import pandas as pd

path = "db/data.xlsx"
sheet = "Датасет(обновленный)"

raw = pd.read_excel(path, sheet_name=sheet, header=None)

markers = {
    "tasks":        "Tasks (Таблица задач и скоринга)",
    "spent_roles":  "Spent_time_roles.",
    "estimates":    "Estimates.",
    "dependencies": "Task_dependencies",
    "engineers":    "Engineers_profiles",
    "team_history": "Team_history",
}

# индекс строки-маркера
marker_idx = {}
for key, text in markers.items():
    hits = raw.index[raw[0].astype(str).str.strip().str.startswith(text)].tolist()
    marker_idx[key] = hits[0]

# для каждого маркера ищем первую непустую строку после него — это заголовки
def find_header(start):
    for i in range(start + 1, len(raw)):
        if raw.iloc[i].notna().sum() >= 3:
            return i
    return start + 1

header_idx = {k: find_header(v) for k, v in marker_idx.items()}

order = ["tasks", "spent_roles", "estimates", "dependencies", "engineers", "team_history"]
bounds = [(header_idx[k], header_idx[order[i+1]] if i+1 < len(order) else len(raw))
          for i, k in enumerate(order)]

def make_block(start, end):
    block = raw.iloc[start:end].dropna(how="all")
    block.columns = block.iloc[0]
    block = block[1:].reset_index(drop=True)
    block = block.loc[:, ~block.columns.astype(str).str.startswith("Unnamed")]
    block = block.dropna(axis=1, how="all")
    return block

tasks        = make_block(*bounds[0])
spent_roles  = make_block(*bounds[1])
estimates    = make_block(*bounds[2])
dependencies = make_block(*bounds[3])
engineers    = make_block(*bounds[4])
team_history = make_block(*bounds[5])
```
